### Question 1
The following is a pseudocode algorithm for the `quick sort` function.

The function takes in an array and returns a sorted array in **ascending** order. The algorithm works for most input but is missing out on a few edge cases.
```
FUNCTION quicksort(array)
  if array length = 1
    return array
  select and remove a pivot value from array
  create empty subarrays less and greater
  for each item in array
    if item < pivot
      append item to less
    else
      append item to greater
  return quicksort(less) + pivot + quicksort(greater)
ENDFUNCTION
```
Writ program code for this algorithm including all the amendments you would make to make the function works for **all** cases. Use the sample data available from text file `NAME.txt` and paste this into your program code.

````python
def quicksort(array):
    leng = len(array)
    if leng <= 1:
        return array

    pivot = array[0]
    less = []
    greater = []

    for idx in range(1, leng): #Since we are taking the first element(idx 0) as the pivot, we start from idx 1
        if array[idx] < pivot:
            less.append(array[idx])
        else:
            greater.append(array[idx])

    return quicksort(less) + [pivot] + quicksort(greater)
    
````

````python
file = open('NAME.txt', 'r')
    
names = file.read().split('\n')

cleaned_names = []
#cleaning the data from the txt file
for name in names: 
    cleaned_names.append(name.split(' = ')[1][1:-1])
print(cleaned_names)

print(quicksort(cleaned_names))

file.close()
````

### Question 2
You are provided with the registration dates and email addresses of participants who
registered for an event in two files, `A.txt` and `B.txt`, with key field values in
**ascending** order. The two files are to be `merged` to form a single file, `C.txt`, containing
the same records with the key field values in **ascending** order.

If a participant registered *more than once* having registration record in both files, take
the **latest registration** date as the final entry. 

Write program code to `remove duplicate` entries and `carry out the merge`. The program should also output the `number of duplicate`
entries removed.

````python
def merge(arr):
    length = len(arr)

    if length <= 1:
        return arr

    mid = length // 2
    left = arr[:mid]
    right = arr[mid+1:]

    merge(left)
    merge(right)

    l = r = i = 0
    while l < len(left) and r < len(right):
        if left[l] < right[r]:
            arr[i] = left[l]
            l += 1

        else:
            arr[i] = right[r]
            r += 1

        i += 1

    while l < len(left):
        arr[i] = left[l]
        l += 1
        i += 1

    while r < len(right):
        arr[i] = right[r]
        r += 1
        i += 1
    
    return arr
````

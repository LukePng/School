### Question 1
For the following array `A`, show A after each pass using `bubble sort` and `insertion sort` to arrange the elements in ascending order
|    i  | 0  | 1  | 2  | 3  | 4  | 5  |
|-------|----|----|----|----|----|----|
| A[i]  | 30 | 50 | 70 | 10 | 40 | 60 |

````python
def bubble_sort(arr):
    leng = len(arr)
    last_swap = leng
    while last_swap > 0:
        check = last_swap
        last_swap = 0
        for i in range(check - 1):
            if arr[i] > arr[i+1]:
                arr[i], arr[i+1] = arr[i+1], arr[i]
                last_swap = i+1
        print(arr)
        
````

````python
def insert_sort(arr):
    leng = len(arr)
    for i in range(leng):
        temp = arr[i]

        space = i
        while (space > 0) and (arr[space-1] > temp):
            arr[space] = arr[space-1]
            space -= 1
            
        arr[space] = temp 
        print(arr)
````

````python
A = [30, 50, 70, 10, 40, 60]
B = A.copy()
print("Bubble Sort:")
bubble_sort(A)

print("\nInsertion Sort:")
insert_sort(B)
````

### Question 3
Data is stored in the array `NameList[1:10]`. This data is to be sorted in **ascending** order.
Complete the pseudocode algorithm for an insertion sort.

```python
FOR ThisPointer <- 2 TO 10
    // use a temporary variable to store item which is to be inserted into its correct location
    Temp <- NameList[ThisPointer]
    Pointer <- ThisPointer – 1
    WHILE (Pointer > 1) AND (NameList[Pointer] > Temp)
        // move list item to next location
        NameList[Pointer] <- NameList[Pointer - 1]
        Pointer <- Pointer - 1
    ENDWHILE
    // insert value of Temp in correct location
    NameList[Pointer] <- Temp
ENDFOR
```

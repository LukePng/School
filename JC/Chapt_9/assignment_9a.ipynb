### Question 1
We want to use linear search to count the number of occurrences of a key in a list. 
When the key does not exist in the list, it is 0 occurrence. 

Write a recursive function `RecSearch(A,key,start)` which requires three parameters, 
the list A, the key to be found, and the start index to search in the list. 

This function returns the number of occurrences of this key in the list.

````python
def RecSearch(A, key, start):
    length = len(A)

    if length < start:
        print("Error: Starting position exceeds length of list")

    num_occ = 0

    for elem in A[start:]:
        if elem == key:
            num_occ += 1

    return num_occ
````

### Question 2
Write a recursive function `BinSearch(A,low,high,key)` 

which searches a sorted list A, A[low] . . . . . A[high], 

for a match with key using binary search. Return the index of the matching item or −1 if the key is not found.

````python
def BinSearch(A, low, high, key):
    if low > high:
        return -1
    
    mid = (low + high) // 2

    if A[mid] == key:
        return mid
    
    elif A[mid] > key:
        return BinSearch(A, low, mid-1, key)
    
    else:
        return BinSearch(A, mid+1, high, key)
````

### Question 3
Customers are identified by ID numbers. These ID numbers are to be stored in a hash table. 

The hashing function to be used is

```Address <--- IDnumber MOD Max```

The hash table is implemented as a one-dimensional array with elements indexed 0 to (Max-1)

#### Task 1
Write program code to:
* Read ID numbers from a text file and store them in a hash table. For the purpose of testing the program, Max is to be set to the value 20. Assume different IDs will hash to different addresses (no collisions).
* Print out the contents of the hash table in the order in which the elements are stored in the array.

Use `KEYS1.TXT` to test your program code.

````python
f1 = open('KEYS1.TXT', 'r')
hash_table = [' ' for _ in range(20)]

#Assuming no collision
for id in f1.readlines():
    try:
        id = int(id)
        idx = id % 20
        hash_table[idx] = id

    except:
        continue
    
print(hash_table)
f1.close()
````

#### Task 2
Amend your program code so that collisions can be managed using open hashing.
This means a collision is resolved by searching sequentially from the hashed address for an empty location and storing the ID at this empty location.

Use `KEYS2.TXT` to test your program code.


````python
f2 = open('KEYS2.TXT', 'r')
hash_table = [' ' for _ in range(20)]

#Assuming no collision
for id in f2.readlines():
    try:
        id = int(id)
        idx = id % 20
        while hash_table[idx] != ' ':
            idx += 1
            if idx == 19:
                idx = 0

        hash_table[idx] = id

    except:
        continue
    
print(hash_table)
f2.close()
````

#### Task 3
Add code to your Task 2 program.

The program is to:
* Take as input an ID number
* Search the hash table and output the address (index number) of the hash table where the ID was found.

Use `KEYS2.TXT` to test your program code.

Run the program three times. Use the following inputs: 37, 77 and 97.

````python
def search_hash(id, hash_table):
    length = len(hash_table)

    for idx in range(length):
        if hash_table[idx] == id:
            return idx
        
    return -1
````

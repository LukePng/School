```python
# question 3
# For each sequence of characters, draw the binary search tree and traverse the tree using
# inorder, preorder, and postorder.
# (a) M, T, V, F, U, N
# (b) F, L, O, R, I, D, A

class TreeNode:
    def __init__(self, data, left = None, right = None):
        self.data = data
        self.left = left # left child
        self.right = right # right child

class BSTree:
    def __init__(self, root = None):
        self.root = root

    def insert(self, new_data):
        new_node = TreeNode(new_data)

        if self.root == None:
            self.root = new_node
            return
        
        # travesal
        inserted = False
        ptr = self.root

        while not inserted:
            if new_data > ptr.data:
                if ptr.right is None:
                    ptr.right = new_node
                    inserted = True
                else:
                    ptr = ptr.right
            
            else:
                if ptr.left is None:
                    ptr.left = new_node
                    inserted = True
                else:
                    ptr = ptr.left

    def inorder_rec(self, root): # in numerical order
        if root:
            self.inorder_rec(root.left)
            print(root.data)
            self.inorder_rec(root.right)
 
    def preorder_rec(self, root): 
        if root:
            print(root.data)
            self.preorder_rec(root.left)
            self.preorder_rec(root.right)

    def postorder_rec(self, root):
        if root:
            self.postorder_rec(root.left)
            self.postorder_rec(root.right)
            print(root.data)
        

Tree_1 = BSTree()
Tree_2 = BSTree()
a = ['M', 'T', 'V', 'F', 'U', 'N']
b = ['F', 'L', 'O', 'R', 'I', 'D', 'A']
for i in a:
    Tree_1.insert(i)

for i in b:
    Tree_2.insert(i)

print("Inorder")
Tree_2.inorder_rec(Tree_2.root)
print("Preorder")
Tree_2.preorder_rec(Tree_2.root)
print("Postorder")
Tree_2.postorder_rec(Tree_2.root)
print()
print("Inorder")
Tree_1.inorder_rec(Tree_1.root)
print("Preorder")
Tree_1.preorder_rec(Tree_1.root)
print("Postorder")
Tree_1.postorder_rec(Tree_1.root)
```

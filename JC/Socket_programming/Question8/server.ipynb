```python
import socket
```

```python
p_socket = socket.socket()
p_socket.bind(("127.0.0.1", 12345))
p_socket.listen()
```

```python
new_socket, addr = p_socket.accept()
print("connected to", str(addr))
```

```python
data = b''

while True:
    msg = str(input("Enter what you want to send to the server"))
    
    new_socket.sendall(msg.encode())
    
    if msg == 'quit':
        break

    data = new_socket.recv(1024)
    if data == b'quit':
        break

    print(data)

new_socket.close()
p_socket.close()
```

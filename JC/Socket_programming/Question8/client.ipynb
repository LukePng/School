```python
import socket
```

```python
client_socket = socket.socket()
client_socket.connect(("122.11.245.128", 12345))
```

```python
data = b""

while True:
    data = client_socket.recv(1024)
    

    print(data)

    if data == b'quit':
        break

    msg = str(input("Enter a message to send to the server: "))
    if msg == 'quit':
        break

    client_socket.sendall(msg.encode())

client_socket.close()
```

#### Insert a new document using db.collection.insert_one()

The insert_one() method inserts a new document into the collection and returns an instance of InsertOneResult, which has a property, inserted_id, that holds the id of the inserted document.

```

```

#### Update one document using db.collection.update_one()

The update_one() method updates the first document that satisfies the query in a collection.

```
    myquery = {"address": "Valley 345"}
    newval = {"$set":{"address":"Canyon 123"}}
    
    db = client.mydatabase
    result = db.customers.update_one(myquery, newval)
   
    for document in db.customers.find():
        print(document)
```
    

#### Update many documents using db.collection.update_many()

The update_many() method updates all documents that satisfy the query in a collection.

```
    db = client.mydatabase
    myquery = {"name": "Minnies"}
    newval = {"$set":{"name":"Minnie"}}
    
    result = db.customers.update_many(myquery, newval)
    print(f"No of documents updated : {result.modified_count}")

```

#### Task 8

List the number of documents without isbn field or document with pageCount=0 in the books collection.
List the total number of documents in the books collection.

```
Documents without isbn or with page count=0 :167
Total number of documents in the collection: 438
```

````python
query = {"$or": [{"pageCount": 0}, {"isbn": {"$exists": False}}]}

without = books.count_documents(query)
total = books.count_documents({})

print(f"Documents without isbn or with page count=0 :{without}")
print(f"Total number of documents in the collection: {total}")
````

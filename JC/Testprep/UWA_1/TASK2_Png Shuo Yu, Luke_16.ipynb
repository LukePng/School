```python
# Task 2.1

class Array:
    def __init__(self, array, length):
        self.array = array
        self.length = length
        
    def size(self):
        return self.length
    
    def get(self, INTEGER):
        return self.array[INTEGER-1]
    
    def set(self, DATA, INTEGER):
        if INTEGER <= self.size(): # Since no.of nodes and index already defined and fixed
            self.array[INTEGER-1] = DATA 
        else:
            return "INTEGER OUT OF RANGE"

length = 7
nodes = []
for i in range(length):
    nodes.append(None)

arr = Array(nodes, 7)
```

```python
# Task 2.2

class BST(Array):
    def __init__(self, array, length):
        super().__init__(array, length)
    
    def count(self):
        ctr = 0
        for i in range(self.size()):
            if self.get(i) is not None:
                ctr += 1
        return ctr
    
    def insert(self, DATA):
        # Given that the idx of the left subtree of the bst is always x2 of the current idx
        # And the idx of the right subtree is that of the left + 1
        # This implementation will traverse through the array as if it is a BST
        if self.get(1) is None: # If the root of the bst is at idx 1
            self.set(DATA, 1)
            return
            
        curr = 0
        left = (curr) * 2 + 1
        right = left + 1
        
        
        while self.get(curr+1) is not None:
            if self.get(curr+1) < DATA:
                if self.get(right+1) is None:
                    self.set(DATA, right+1)
                    print(f'INSERTED! AT NODE{right  + 1}')
                    return
                
                else:
                    curr = right
            
            else: 
                if self.get(left+1) is None:
                    self.set(DATA, left+1)
                    print(f'INSERTED! AT NODE{left  + 1}')
                    return
                
                else:
                    curr = left
            left = (curr) * 2 + 1
            right = left + 1

# Test Case
# tree1 = BST(nodes, length)
# tree1.insert(2)
# tree1.insert(1)
# tree1.insert(3)
# tree1.insert(0)
# tree1.get(4)
```

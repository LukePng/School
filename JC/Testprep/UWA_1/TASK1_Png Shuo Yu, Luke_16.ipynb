```python
# Task 1.1

def task1_1(filename):
    names = open(filename, 'r')
    namelist = []
    for name in names.readlines():
        cleaned = name.split('\n')
        namelist.append(cleaned[0])
    
    names.close()
    return namelist

names = task1_1('namelist.txt')
print(names[:10])
print(len(names))
```

```python
# Task 1.3

def quick(arr, left, right):
    if left < right:
        pivot = arr[left][0]
        space = left
        
        for i in range(left+1, right+1):
            if arr[i][0] < pivot:
                space += 1
                arr[i], arr[space] = arr[space], arr[i]
                
        arr[space], arr[left] = arr[left], arr[space]
    
        quick(arr, left, space-1)
        quick(arr, space+1, right)
                

def task1_3(list_of_tuples):
    right = len(list_of_tuples)
    quick(list_of_tuples, 0, right-1)
    return list_of_tuples

# Test_cases
# sorted_list = task1_3(lst_of_tuples)
# sorted_list
```

```python
# Task 1.4

def task1_4(sorted_list, name):
    left = 0
    right = len(sorted_list) - 1
    
    while left <= right:
        mid = (left + right) // 2
        
        if sorted_list[mid][0] == name:
            return sorted_list[mid][1]
        
        elif name < sorted_list[mid][0]:
            right = mid - 1
        
        else:
            left = mid + 1
            
    return -1

# Test Cases
# yes_case = task1_4(sorted_list, 'ZACHARIAS PADILLA')
# no_case = task1_4(sorted_list, 'helpme')
# print(yes_case)
# print(no_case)
```

```python
# Task 1.5

def task1_5(filename_in, filename_out):
    names = task1_1(filename_in)
    lst_of_tuples = task1_2(names)
    sorted_names = task1_3(lst_of_tuples)
    
    output = open(filename_out, 'w')
    
    output.write('Name, Table\n')
    for name_table in sorted_names:
        output.write(f'{name_table[0]}, {name_table[1]}\n')
        
    output.close()

    
task1_5('namelist.txt', 'table_allocation.csv')
```
